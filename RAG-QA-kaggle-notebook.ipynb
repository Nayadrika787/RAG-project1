```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("/kaggle/input/datasets/nayanasen/dbmstxt2/dbms.txt")
documents = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(documents)
```

```python
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embedding_model = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
db=Chroma.from_documents(chunks,embedding_model)
```

```python
retriever = db.as_retriever(search_kwargs={"k": 3})
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch
```

```python
#model_name = "mistralai/Mistral-7B-Instruct-v0.1"
model_name = "google/gemma-2b-it" 
# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32, # Use 'dtype' instead of deprecated 'torch_dtype'
    device_map="auto"      #Automatically places model on CPU/GPU
)

# Create the text generation pipeline
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)
```

```python
def question(query):
    docs = retriever.invoke(query)
    
    # Raw answer (accurate)
    raw = docs[0].page_content
    # Clean answer (LLM)
    context = " ".join([doc.page_content for doc in docs[:3]])

    context = context.replace("\n", " ")    # fix real newline
    context = context.replace("  ", " ")    # remove extra spaces    
    prompt = f"""
    <start_of_turn>user
    Use the following context to answer the question in exactly two sentences.
    
    Context: {context}
    Question: {query}
    <end_of_turn>
    <start_of_turn>model
    """
    
    result = generator(
        prompt,
        max_new_tokens=100,
        return_full_text=False,
        pad_token_id=tokenizer.eos_token_id,
        do_sample=False,
        temperature=0.0,
        top_p=0.95,
    )
    full_text = result[0].get('generated_text', '')
    clean = full_text.strip()
    return clean
```

```python
print(question("What is a relationship set"))
```
